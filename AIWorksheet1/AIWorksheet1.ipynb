```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
DRIVE_PATH = '/content/drive/MyDrive/worksheet1/image.png'
```

```python
#read and display the image
img = Image.open(DRIVE_PATH)
img_array = np.array(img)
plt.figure(figsize=(15, 10))
plt.subplot(2, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis('off')
```

```python
#displaying top left corner of 100x100 pixels
top_left = img_array[0:100, 0:100]
plt.subplot(2, 3, 2)
plt.imshow(top_left)
plt.title("Top Left Corner (100x100)")
plt.axis('off')
```

```python
#displaying three color channels (R, G, B)
R = img_array[:, :, 0]
G = img_array[:, :, 1]
B = img_array[:, :, 2]
plt.figure(figsize=(15, 5))

#red channel
plt.subplot(1, 3, 1)
plt.imshow(R, cmap='Reds')
plt.title("Red Channel")
plt.axis('off')

#green channel
plt.subplot(1, 3, 2)
plt.imshow(G, cmap='Greens')
plt.title("Green Channel")
plt.axis('off')

#blue channel
plt.subplot(1, 3, 3)
plt.imshow(B, cmap='Blues')
plt.title("Blue Channel")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
#top 100x100 pixels to a value of 210
modified_img = img_array.copy()
modified_img[0:100, 0:100] = 210
plt.subplot(2, 3, 6)
plt.imshow(modified_img)
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
DRIVE_PATH_EX2 = '/content/drive/MyDrive/worksheet1/image1.png'

#displaying grayscale image
img_gray = Image.open(DRIVE_PATH_EX2).convert('L')
img_gray_array = np.array(img_gray)
plt.figure(figsize=(18, 10))
plt.subplot(2, 3, 1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')
```

```python
#extract and display the middle section (150x150 pixels)
h, w = img_gray_array.shape
center_h, center_w = h // 2, w // 2
half_size = 150 // 2

middle_section = img_gray_array[
    center_h - half_size : center_h + half_size,
    center_w - half_size : center_w + half_size
]

plt.subplot(2, 3, 2)
plt.imshow(middle_section, cmap='gray')
plt.title("Middle 150x150 Pixels")
plt.axis('off')
```

```python
#applying simple threshold (100)
#pixels < 100 become 0 others become 255 to create a binary image
thresholded = np.where(img_gray_array < 100, 0, 255)
plt.subplot(2, 3, 3)
plt.imshow(thresholded, cmap='gray')
plt.title("Thresholded Image")
plt.axis('off')
```

```python
#rotating the image 90 degrees clockwise
rotated_img = img_gray.rotate(-90, expand=True)
plt.subplot(2, 3, 4)
plt.imshow(rotated_img, cmap='gray')
plt.title("Rotated Image (90° CW)")
plt.axis('off')
```

```python
#convert grayscale to RGB
rgb_converted = np.stack([img_gray_array] * 3, axis=-1)
plt.subplot(2, 3, 5)
plt.imshow(rgb_converted)
plt.title("Converted RGB Image")
plt.axis('off')
plt.tight_layout()
plt.show()
```
