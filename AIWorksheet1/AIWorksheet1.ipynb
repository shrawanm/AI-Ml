```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
DRIVE_PATH_EX2 = '/content/drive/MyDrive/worksheet1/image1.png'
img_gray = Image.open(DRIVE_PATH_EX2).convert('L')
img_gray_array = np.array(img_gray)
plt.figure(figsize=(18, 10))
plt.subplot(2, 3, 1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')
```

```python
#extract and display the middle section (150x150 pixels)
h, w = img_gray_array.shape
center_h, center_w = h // 2, w // 2
half_size = 150 // 2

middle_section = img_gray_array[
    center_h - half_size : center_h + half_size,
    center_w - half_size : center_w + half_size
]

plt.subplot(2, 3, 2)
plt.imshow(middle_section, cmap='gray')
plt.title("Middle 150x150 Pixels")
plt.axis('off')
```

```python
thresholded = np.where(img_gray_array < 100, 0, 255)
plt.subplot(2, 3, 3)
plt.imshow(thresholded, cmap='gray')
plt.title("Thresholded Image")
plt.axis('off')
```

```python
#rotating the image 90 degrees clockwise
rotated_img = img_gray.rotate(-90, expand=True)
plt.subplot(2, 3, 4)
plt.imshow(rotated_img, cmap='gray')
plt.title("Rotated Image (90° CW)")
plt.axis('off')
```

```python
#convert grayscale to RGB
rgb_converted = np.stack([img_gray_array] * 3, axis=-1)
plt.subplot(2, 3, 5)
plt.imshow(rgb_converted)
plt.title("Converted RGB Image")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

DRIVE_PATH = '/content/drive/MyDrive/worksheet1/image2.png'
img = Image.open(DRIVE_PATH).convert('L')
img_array = np.array(img, dtype=float)

#subtract mean of each column
mean = np.mean(img_array, axis=0)
img_centered = img_array - mean

#calculate the covariance matrix
cov_matrix = np.cov(img_centered, rowvar=False)

#eigen Decomposition and Identifying Principal Components
eigen_values, eigen_vectors = np.linalg.eigh(cov_matrix)

#sort eigenvalues and eigenvectors in descending order
sorted_index = np.argsort(eigen_values)[::-1]
sorted_eigenvalue = eigen_values[sorted_index]
sorted_eigenvectors = eigen_vectors[:, sorted_index]

# Identify Principal Components with Cumulative Sum plot
explained_variance = sorted_eigenvalue / np.sum(sorted_eigenvalue)
cumulative_variance = np.cumsum(explained_variance)

plt.figure(figsize=(8, 5))
plt.plot(cumulative_variance, marker='o', linestyle='--', color='b')
plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.grid()
plt.show()

def reconstruct_image(data_centered, mean_val, vectors, k):
    selected_vectors = vectors[:, :k]
    transformed_data = np.dot(data_centered, selected_vectors)
    reconstructed_data = np.dot(transformed_data, selected_vectors.T) + mean_val
    return reconstructed_data
k_values = [10, 20, 50, 100, 150]
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(img_array, cmap='gray')
plt.title("Original Image")
plt.axis('off')

for i, k in enumerate(k_values, 2):
    reconstructed = reconstruct_image(img_centered, mean, sorted_eigenvectors, k)
    plt.subplot(2, 3, i)
    plt.imshow(reconstructed, cmap='gray')
    plt.title(f"{k} Components\n(Var: {cumulative_variance[k-1]:.2%})")
    plt.axis('off')
plt.tight_layout()
plt.show()
```

The exercise demonstrates that image quality improves as more components are retained, transitioning from a blurred 10 component to a sharp 150 component reconstruction that mirrors the original. By calculating the covariance matrix and eigenvalues the cumulative explained variance plot reveals an elbow where a small fraction of components captures the majority of the image data. This evaluation confirms that PCA effectively achieves dimensionality reduction by preserving high-variance structural features while discarding redundant information, allowing for significant data compression with minimal loss in visual fidelity
